```python
from mario_gpt.prompter import Prompter
from mario_gpt import MarioDataset, MarioLM
from transformers import AutoTokenizer
from mario_gpt.utils import view_level

mario_lm = MarioLM()

dataset = MarioDataset(mario_lm.tokenizer,
                       folder_path="mario_gpt/levels"  # Replace with actual path to Mario level data
                   )

view_level(dataset.input_ids[:700], mario_lm.tokenizer)
```

```python
tokenizer_path = "shyamsn97/Mario-GPT2-700-context-length"
level_tokenizer = AutoTokenizer.from_pretrained(tokenizer_path)

prompter = Prompter(level_tokenizer=level_tokenizer)
```

```python
# Test with a level from dataset
print("\nTesting with level from dataset:")
prompt, prompt_dict, str_level = prompter(level=dataset.input_ids[0:700])
print(f"Full prompt: {prompt}")
```
